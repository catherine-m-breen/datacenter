```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import contextily as cx
import numpy as np
from matplotlib_scalebar.scalebar import ScaleBar
import pandas as pd

# /Users/cmbreen/Documents/datacenters/im3_open_source_data_center_atlas_v2026.02.09.csv


locations = '/Users/cmbreen/Documents/datacenters/im3_open_source_data_center_atlas_v2026.02.09.csv'
locations = pd.read_csv(locations)
x, y = locations['lon'], locations['lat']


# Create a new figure
fig, ax = plt.subplots(figsize=(10, 10))

# 1. Set the exact same map boundaries as your previous plot
ax.set_xlim(-79, -76)
ax.set_ylim(37, 41)

# 2. Plot all the data center locations
ax.scatter(x, y, alpha=0.5, color='blue', s=15, label='Data Centers')

# 3. Draw the Red Bounding Box
xmin, ymin, xmax, ymax = -78.0, 38.5, -77.0, 39.5
rect = patches.Rectangle(
    (xmin, ymin), xmax - xmin, ymax - ymin, 
    linewidth=2, edgecolor='red', facecolor='none'
)
ax.add_patch(rect)

# 4. Add specific cities
ax.scatter(-77.0369, 38.8950, color='purple', s=50, zorder=5) # DC
ax.scatter(-77.56, 39.11, color='purple', s=50, zorder=5) # Leesburg

ax.annotate('Washington D.C.', (-77.0369, 38.8950), xytext=(5, 5), 
            textcoords='offset points', color='purple', fontweight='bold', 
            bbox=dict(facecolor='white', alpha=0.6, edgecolor='none', pad=1))
ax.annotate('Leesburg', (-77.56, 39.11), xytext=(5, 5), 
            textcoords='offset points', color='purple', fontweight='bold',
            bbox=dict(facecolor='white', alpha=0.6, edgecolor='none', pad=1))

# 5. ADD THE OPENSTREETMAP BASEMAP
# crs="EPSG:4326" tells contextily that our coordinates are standard lat/lon degrees
cx.add_basemap(ax, crs="EPSG:4326", source=cx.providers.OpenStreetMap.Mapnik)

# 6. Formatting
ax.set_title("Data Center Infrastructure (OpenStreetMap)")
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')

# Add Scale bar
center_lat = 39.0
meters_per_degree = np.cos(np.radians(center_lat)) * 111320.0
scalebar = ScaleBar(dx=meters_per_degree, units="m", length_fraction=0.2, 
                    location="lower right", box_alpha=0.8)
ax.add_artist(scalebar)

plt.tight_layout()
plt.show()
```
